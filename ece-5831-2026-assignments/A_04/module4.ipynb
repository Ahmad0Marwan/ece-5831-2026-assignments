# Module 4 Multilayer Perceptron

This implementation follows the NumPy class and network dictionary approach in the supplied reference. The required architecture is 3 → 5 → 5 → 2 → 2.


```python
import numpy as np
from multilayer_perceptron import MultilayerPerceptron

np.set_printoptions(precision=6, suppress=True)
mlp = MultilayerPerceptron()
```

Network dimensions

```python
for i in range(1, 5):
    activation = "Sigmoid" if i < 4 else "ReLU"
    print(f"Layer {i}: weights {mlp.network[f'w{i}'].shape}, "
          f"biases {mlp.network[f'b{i}'].shape}, {activation}")
```

forward propagation

```python
x = np.array([1.0, 0.5, -1.0])
a1, z1, a2, z2, a3, z3, a4, y = mlp.forward(x, return_layers=True)
print("Input:", x)
for i, (a, z) in enumerate([(a1, z1), (a2, z2), (a3, z3), (a4, y)], start=1):
    activation = "sigmoid" if i < 4 else "ReLU"
    print(f"Layer {i} before activation:", a)
    print(f"Layer {i} after {activation}:", z)
```
